# Metadata and data quality
Synthetic inline records; no account, CSV download or extra package required.


Author: Isaac K. Nti. See [citation and reuse terms](../CITATION.md).


```
records = [{'id': 1, 'category': 'retail', 'amount': 10}, {'id': 2, 'category': None, 'amount': 20}]
catalog = {key: {'types': sorted({type(r[key]).__name__ for r in records}), 'missing': sum(r[key] is None for r in records)} for key in records[0]}
assert catalog['category']['missing'] == 1
catalog
```
